```python
#import the necessary libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#bring Learning_DataScience/Data_Science1(Netflix popular movies dataset)/n_movies_cleaned.csv into the notebook
df = pd.read_csv('n_movies_cleaned.csv')
df.head()
```

```python
#relation between votes(popularity) and rating
plt.scatter(df['votes'], df['rating'])
plt.title('Votes vs Rating')
plt.xlabel('Votes')
plt.ylabel('Rating')
plt.show()
```

```python
# genre explosion
df_exploded = df.assign(genre=df['genre'].str.split(',')).explode('genre')

# clean whitespace
df_exploded['genre'] = df_exploded['genre'].str.strip()
print(df_exploded.head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Average rating by genre
plt.figure(figsize=(12, 6))
sns.barplot(x='genre', y='rating', data=df_exploded, ci=None, order=df_exploded.groupby('genre')['rating'].mean().sort_values(ascending=False).index)
plt.title('Average Rating by Genre')
plt.xlabel('Genre')
plt.ylabel('Average Rating')
plt.xticks(rotation=45)
plt.show()

# Average votes by genre
plt.figure(figsize=(12, 6))
sns.barplot(x='genre', y='votes', data=df_exploded, ci=None, order=df_exploded.groupby('genre')['votes'].mean().sort_values(ascending=False).index)
plt.title('Average Votes by Genre')
plt.xlabel('Genre')
plt.ylabel('Average Votes')
plt.xticks(rotation=45)
plt.show()
```

```python
#relation between duration(runtime) and rating
#round up the value of 'duration'

rounded_duration = df['duration'].round(-1) #round to nearest 10
print(rounded_duration.head())
```

```python
#Now, plot the relationship between rounded duration and rating
plt.scatter(rounded_duration, df['rating'])
plt.title('Duration vs Rating')
plt.xlabel('Duration (rounded to nearest 10 minutes)')
plt.ylabel('Rating')
plt.show()
```
